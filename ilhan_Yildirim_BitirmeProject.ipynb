## Brüt Gelir Sütunu ve Satış Doğrulaması

**Amaç:** Sales sütununun nasıl hesaplandığını doğrulamak.

**Yöntem:** Ürün fiyatı (Product Price) ile alınan adet (Order Item Quantity) çarpılarak Gross Revenue (brüt gelir) adında yeni bir sütun oluşturuldu. Bu sütun, Sales sütunuyla bütün satırlarda karşılaştırıldı.

**Sonuç:** İki sütun satırların %100'inde aynı değeri veriyor. Bu sonuç, Sales sütununun indirim uygulanmadan önceki brüt satış tutarını (fiyat × adet) gösterdiğini doğruluyor.

**Neden önemli:** Kârlılık analizinde satış tutarı olarak Sales değil, indirim sonrası ödenen tutarı gösteren Order Item Total kullanıldı. Bu kontrol, iki sütun arasındaki farkın indirimden kaynaklandığını ve doğru sütunun seçildiğini destekliyor.

```python
df["Gross Revenue"] = df["Product Price"] * df["Order Item Quantity"]

print(df[["Product Price", "Order Item Quantity", "Gross Revenue", "Sales"]].head())
```
