## Yeni Görevler

Bir önceki adımdaki sildiğimiz 2 sütunu dataco_Proje.csv şeklinde kaydettim ve veri analizine bu dosyadan devam ediyorum.



```python
import pandas as pd 

df = pd.read_csv("dataco_Proje.csv", encoding="latin-1")

# Metin formatı UTF-8 değil , bu yüzden latin-1 kullandım.
```

```python
df.describe()
```

```python
sale = df["Sales"].mean()
print(sale)
```

```python
sale_m = df["Sales"].median()
print(sale_m)
```

```python
sale_1 = df["Sales"].min()
print(sale_1)
```

```python
sale_2 = df["Sales"].max()
print(sale_2)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

Q1 = df["Sales"].quantile(0.25)
Q3 = df["Sales"].quantile(0.75)
IQR = Q3 - Q1

alt_sinir = Q1 - 1.5 * IQR
ust_sinir = Q3 + 1.5 * IQR

aykiri_degerler = df[(df["Sales"] < alt_sinir) | (df["Sales"] > ust_sinir)]

print(f"Alt Sınır: {alt_sinir:.2f}")
print(f"Üst Sınır: {ust_sinir:.2f}")
print(f"Aykırı Değer Sayısı: {len(aykiri_degerler)}")
print(f"Veri Setine Oranı: %{(len(aykiri_degerler) / len(df)) * 100:.2f}")

```

```python
print("Hatalı Satış tutarı :", (df["Sales"] <= 0).sum())
print("Hatalı Adet sayısı :", (df["Order Item Quantity"] <= 0).sum())
print("Hatalı Ürün Fiyati :", (df["Product Price"] <= 0).sum())
```
