```python
import pandas as pd 

df = pd.read_csv("DataCoSupplyChainDataset.csv", encoding="latin-1")

# Metin formatı UTF-8 değil , bu yüzden latin-1 kullandım.
```

## Frekans Bulma : hangi müşteri segmenti kaç sipariş kalemi vermiş sorusunun yanıtını buluyoruz.


```python
print(df["Customer Segment"].value_counts())
```

#  Sıralama: en çok kâr getiren 5 sipariş kalemi


```python
en_karli_5 = df.sort_values("Order Profit Per Order", ascending=False).head(5)

print(en_karli_5[["Order Id", "Product Name", "Order Item Quantity", "Order Item Total", "Order Profit Per Order"]])
```

#  Gruplama: her müşteri segmentinin toplam satışı ve toplam kârı


```python
segment_satis = df.groupby("Customer Segment")["Order Item Total"].sum()
segment_kar = df.groupby("Customer Segment")["Order Profit Per Order"].sum()

print(segment_satis.round(2).sort_values(ascending=False))
print(segment_kar.round(2).sort_values(ascending=False))
```

```python
segment_kar = df.groupby("Customer Segment")["Order Profit Per Order"].sum()

print(segment_kar.reset_index())   
```

```python
print(df.groupby("Customer Segment")["Order Profit Per Order"].agg(["sum", "mean", "count"]))
```
