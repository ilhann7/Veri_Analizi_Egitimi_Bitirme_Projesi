## Boş Olan Sütun Silme İşlemi

**1** 'Order Zipcode' boş olan bir sütundu ve silindi. 

```python
clear_df = df
clear_df = df.drop(columns=["Order Zipcode"])
```

```python
clear_df.info()
```

```python
clear_df.shape
```

**1** ) Görüldüğü üzere sütun sayımız 53'den 52'ye düştü.

```python
clear_df = clear_df.drop(columns=["Product Description"])
```

```python
clear_df.shape
```

**1** ) Customer Zipcode sütununun modunu buluruz ve boşlukları mod(en çok tekrar eden) değeriyle doldururuz.

```python
zipcode_mod = clear_df["Customer Zipcode"].mode()[0]
clear_df["Customer Zipcode"] = clear_df["Customer Zipcode"].fillna(zipcode_mod)
```

Boş olan satırlar "Unknown" ile dolduruldu.

```python
clear_df["Customer Lname"] = clear_df["Customer Lname"].fillna("Unknown")
```

```python
clear_df.info()
```
